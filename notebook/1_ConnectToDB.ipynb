# Connect to the database
First import the needed dependencies

```python
import sqlite3
```

Define a function for pinging the db. This is not needed in go

```python
def ping_db(conn) -> bool:
    try:
        cursor = conn.cursor()
        cursor.execute("SELECT 1")
        cursor.fetchone()
        print("Database connection is active.")
        return True
    except sqlite3.Error as e:
        print(f"Database connection failed: {e}")
        return False
```

connect to the db and confirm connected

```python
connection = sqlite3.connect("file::memory:?cache=shared")
assert ping_db(connection)  # noqa: F821 (ping_db is defined above)
connection.close() # * Just some clean up
```
